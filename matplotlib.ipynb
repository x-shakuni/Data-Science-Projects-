```python
# types of plot
# a. Linear Plot
# b. Scatter Plot
# c. Bar Plot
# d. Stem Plot
# e. Step Plot
# f. Hist Plot
# g. Box Plot
# h. Pie Plot
# i. Fill_Between Plot
```

```python
import matplotlib.pyplot as plt
```

```python
# Linear graph
```

```python
a = [1,2,3,4]
b = [5,6,7,8]
plt.plot(a,b)
plt.show()
```

```python
a = [1,2,3,4]
b = [5,6,7,8]
c = 'r'
plt.plot(a,b,c)
plt.show()
```

```python
x = [5,2,9,4,7]
y = [10,5,8,4,2]

plt.plot(x,y, marker = 'o')
```

```python
x = [5,2,9,4,7]
y = [10,5,8,4,2]

plt.plot(x,y)
plt.plot(y, marker = '>')
plt.plot(x, marker = 'o')
```

```python
# Bar graph
```

```python
x = [1,2,3,4]
y = [5,6,7,8]

plt.bar(x,y)
plt.show()
```

```python
x = [1,2,3,4]
y = [5,6,7,8]
c = ['r', 'g', 'b', 'y']
plt.bar(x,y, color = c)
plt.show()
```

```python
import numpy as np
```

```python
x = np.array(['A', 'B', 'C', 'D'])
y = np.array([3,8,1,10])

plt.barh(x,y, color = '#1fcd76')
plt.show()
```

```python
 x = np.array(['A', 'B', 'C', 'D'])
y = np.array([3,8,1,10])

plt.bar(x,y, color = '#1fcd76', width = 0.3)
plt.show()
```

```python
x = np.array(['A', 'B', 'C', 'D'])
y = np.array([3,8,1,10])

plt.barh(x,y, color = '#1fcd76', height = 0.3)
plt.show()
```

```python
# Scatter Plot
```

```python
x = [1,2,3,4,5,6]
y = [4,2,3,5,1,8]
c = ['red', 'magenta', 'green', 'orange', 'blue', 'brown']

plt.title("Scatter Plot")
plt.xlabel('month', color = 'red')
plt.ylabel('numbers', color = 'blue')
plt.scatter(x,y, color = c, s=75)
plt.show()
```

```python
from PIL import Image

fname = r'image.png'

# opening image using PIL

image = Image.open(fname).convert('L') # L or RGB

# mapping image to greyscale

plt.imshow(image, cmap = 'inferno')
# plt.show()
```

```python
# Pie Chart
```

```python
x = [10,20,30,40]
y = ['English', 'Hindi', 'Maths', 'Science']

plt.title("Marks")
plt.pie(x, labels = y)
plt.show()
```

```python
x = [10,20,30,40]
y = ['English', 'Hindi', 'Maths', 'Science']

plt.title("Marks")
plt.pie(x, labels = y)
plt.legend()
plt.show()
plt.savefig()
```
