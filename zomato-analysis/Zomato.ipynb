## Zomato Data Analysis Project

```python
# importing libraries
import pandas as pd # used for data manipulation and analysis
import numpy as np # used for numerical operations
import matplotlib.pyplot as plt # for data visualisation
import seaborn as sns # for data visualisation
```

```python
# creating data frame
dataframe = pd.read_csv("Zomato data .csv")
print(dataframe)
```

```python
dataframe
```

```python
# data cleaning
# convert the data type of column rate
def handleRate(value):
    value = str(value).split('/')
    value = value[0]
    return float(value)

dataframe['rate']= dataframe['rate'].apply(handleRate)
print(dataframe.head())
```

```python
dataframe.info()
```

```python
# type of resturant
dataframe.head()
```

```python
sns.countplot(x=dataframe['listed_in(type)'])
plt.xlabel("Type of Resturant")
```

```python
# votes received by each type of resturant
grouped_data = dataframe.groupby('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes': grouped_data})
plt.plot(result, c="green", marker="0")
plt.xlabel("Type of restaurant", c="red", size=20)
plt.ylabel("Votes", c="red", size=20)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming 'dataframe' is your Pandas DataFrame
grouped_data = dataframe.groupby('listed_in(type)')['votes'].sum()

plt.bar(grouped_data.index, grouped_data.values, color="green")
plt.xlabel("Type of restaurant", c="red", size=20)
plt.ylabel("Votes", c="red", size=20)
plt.title("Total Votes by Restaurant Type")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
