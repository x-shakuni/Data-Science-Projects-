```python
# Different categories of plot in Seaborn
# Plots are basically used for visualizing the relationship between variables. Those variables can
# be either completely numerical or a category like a group, class, or division. Seaborn divides
# the plot into the below categories -
# Relational plots: This plot is used to understand the relation between two variables.
# Categorical plots: This plot deals with categorical variables and how they can be visualized.
# Distribution plots: This plot is used for examining univariate and bivariate distributions
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
var1 = [1,2,3,4,5,6,7,8]
var2 = [8,7,6,5,4,3,2,1]

plt.plot(var1, var2)
plt.show()
```

```python
sns.lineplot(x = var1, y = var2)
plt.show()
```

```python
import pandas as pd
```

```python
df = pd.DataFrame({"var1":var1, "var2":var2})
sns.lineplot(x = 'var1', y = 'var2', data = df)
plt.show()
```

```python
df1 = sns.load_dataset("penguins")
```

```python
df1
```

```python
sns.lineplot(x = 'bill_length_mm', y = 'flipper_length_mm', data = df1, hue = "sex")
plt.show()
```

```python
sns.lineplot(x = 'bill_length_mm', y = 'flipper_length_mm', data = df1, hue = "sex", palette = 'rocket_r', markers = ["o", ">"])
plt.show()
```

```python
df2 = sns.load_dataset("penguins").head(20)
```

```python
df2
```

```python
sns.lineplot(x = 'bill_length_mm', y = 'flipper_length_mm', data = df2, hue = "sex", palette = 'rocket_r', markers = ["o", ">"])
plt.show()
```

```python
markers_dict = {'Male': 'o', 'Female': '>'} # Assuming 'Male' and 'Female' are the values in your 'sex' column
sns.lineplot(x='bill_length_mm', y='flipper_length_mm', data=df2, hue="sex", style="sex", palette='rocket_r', markers=markers_dict)
plt.show()
```

```python
sns.lineplot(x='bill_length_mm', y='flipper_length_mm', data=df2, hue="sex", style="sex", palette='rocket_r', markers=["o", ">"])
plt.show()
```

```python
sns.lineplot(x='bill_length_mm', y='flipper_length_mm', data=df2, hue="sex", style="sex", palette='rocket_r', markers=["o", ">"])
plt.title("This is a line chart related to penguines")
plt.grid()
plt.show()
```
