```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1 = sns.load_dataset("penguins")
```

```python
df1
```

```python
df1["island"]
```

```python
sns.barplot(x = df1.island, y = df1.bill_length_mm)
plt.show()
```

```python
sns.barplot(x = 'island', y = 'bill_length_mm', data = df1, hue = 'sex')
plt.show()
```

```python
## order 
```

```python
order1 = ["Dream", "Torgersen", "Biscoe"]
sns.barplot(x = 'island', y = 'bill_length_mm', data = df1, hue = 'sex', order = order1)
plt.show()
```

```python
order1 = ["Dream", "Torgersen", "Biscoe"]
sns.barplot(x = 'island', y = 'bill_length_mm', data = df1, hue = 'sex', order = order1, hue_order = ["Female", "Male"], palette = 'inferno')
plt.show()
```

```python
# Saturation parameter
```

```python
sns.barplot(x = 'island', y = 'bill_length_mm', data = df1, hue = 'sex', palette = 'inferno', saturation = 2.5)
plt.show()
```
